```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# --- 1. PREPARAÇÃO DOS DADOS ---
dados_categoria = pd.DataFrame({
    'Classificação': ['DII Ped', 'Início precoce', 'VEOIBD'],
    'Pacientes': [39, 14, 47],
    'Porcentagem': [39.0, 14.0, 47.0]
})

# --- 2. CONFIGURAÇÃO VISUAL ---
sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(10, 6))

# --- 3. PLOTAGEM DAS BARRAS ---
# O parâmetro edgecolor='#333333' e linewidth=1.2 cria aquele contorno escuro fininho na barra
barras = sns.barplot(
    data=dados_categoria, 
    x='Classificação', 
    y='Porcentagem', 
    color='#6B9ACD', # O azul exato do seu recorte
    edgecolor='#333333', # Borda escura igual ao seu recorte
    linewidth=1.2,
    width=0.6, 
    ax=ax
)

# --- 4. ADICIONANDO O TEXTO "n = XX" ACIMA DAS BARRAS ---
for i, p in enumerate(ax.patches):
    altura = p.get_height()
    n_pacientes = dados_categoria['Pacientes'].iloc[i]
    
    # Posição X: centro da barra
    x_pos = p.get_x() + p.get_width() / 2
    
    # Posição Y: acima da barra (+ 1.5 para dar um espacinho)
    y_pos = altura + 1.5
    
    # Adiciona o texto no formato exato: "n = 56", preto, centralizado
    ax.text(
        x_pos, 
        y_pos, 
        f"n = {n_pacientes}", 
        ha='center',     
        va='bottom',     # Alinha a base do texto com o y_pos
        fontsize=12,     # Fonte um pouco maior
        color='black',   # Cor preta (igual ao recorte)
        fontfamily='sans-serif'
    )

# --- 5. AJUSTES DE EIXOS E TÍTULOS ---
ax.set_xlabel("Classificação etária", fontsize=12, fontweight='bold')
ax.set_ylabel("Pacientes (%)", fontsize=12, fontweight='bold')

# Formata o eixo Y para mostrar o símbolo de porcentagem
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x:.0f}%'))

# AQUI ESTÁ O SEGREDO: Aumentar o limite do eixo Y para 55% 
# para o texto não encostar na linha do topo do gráfico.
ax.set_ylim(0, 55) 

# Remove as linhas de grade verticais e as bordas superior e direita
ax.grid(axis='x', visible=False)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Deixa as linhas de grade horizontais um pouco mais suaves (cinza claro)
ax.grid(axis='y', color='#E0E0E0', linestyle='-', linewidth=1)

plt.tight_layout()

# --- 6. SALVANDO E EXIBINDO ---
nome_arquivo = "grafico_classificacao_n_acima.png"
plt.savefig(nome_arquivo, dpi=300, bbox_inches='tight', facecolor='white')
print(f"Gráfico salvo com sucesso como: {nome_arquivo}")

# plt.show()

# Ajusta o layout para caber a nota de rodapé
# plt.tight_layout(rect=[0, 0.03, 1, 1]) 

# --- 9. SALVANDO O GRÁFICO (OBRIGATORIAMENTE ANTES DO SHOW) ---
nome_arquivo = "class_etaria_histograma.png" 

plt.savefig(
    nome_arquivo, 
    dpi=300,              # Alta resolução
    bbox_inches='tight',  # Remove espaços em branco
    facecolor='white'     # Fundo branco
)

print(f"Gráfico salvo com sucesso como: {nome_arquivo}")

# --- 10. EXIBINDO O GRÁFICO ---
# O show deve ser o ÚLTIMO comando
plt.show()
```
